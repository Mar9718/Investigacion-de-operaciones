```python
from sympy import *
init_printing(use_unicode=True)
import matplotlib.pyplot as plt
```

```python
P = Matrix([[0.6, 0.4], [0.2, 0.8]])
display(P)
```

```python
pi_0= Matrix([[1, 0]])
display(pi_0)
```

```python
pi_5 = pi_0*P**5
display(pi_5)
```

```python
pi_10 = pi_0*P**100
display(pi_10)
```

```python
dias = range(6)
plt.plot(dias, [float((pi_0 * P**dia)[0]) for dia in dias], marker="o")
plt.xlabel("Día"); plt.ylabel("P(Soleado)"); plt.grid();
# Marca la probabilidad de largo plazo con una línea horizontal.
plt.axhline(
    y=1/3,
    color="red",
    linestyle="--",
    label="Convergencia: 0.3333"
)

plt.title("Convergencia de la probabilidad de un día soleado")
plt.legend()
plt.show()
```

```python
dias = range(31)
plt.plot(dias, [float((pi_0 * P**dia)[1]) for dia in dias], marker="o")
plt.xlabel("Día"); plt.ylabel("P(Lluvioso)"); plt.grid();
# Marca la probabilidad de largo plazo con una línea horizontal.
plt.axhline(
    y=1/3,
    color="red",
    linestyle="--",
    label="Convergencia: 0.6666"
)

plt.title("Convergencia de la probabilidad de un día lluvioso")
plt.legend()
plt.show()
```
